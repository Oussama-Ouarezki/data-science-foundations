```python
import numpy as np
import matplotlib.pyplot as plt

# Hyperparameters
learning_rate = 0.1
epochs = 50

# Initialize parameter
theta = np.random.rand()
theta_history = []  # To store theta values for plotting
loss_history = []   # To store loss values for plotting

# Dataset (single data point example)
x = np.array([1])  # Input (dummy variable)
y = np.array([3])  # Target (optimal theta)

# SGD loop
for epoch in range(epochs):
    # Predict
    prediction = theta
    # Compute loss
    loss = (prediction - y) ** 2
    # Compute gradient
    gradient = 2 * (prediction - y)
    # Update parameter
    theta = theta - learning_rate * gradient
    # Store values for plotting
    theta_history.append(theta)
    loss_history.append(loss)

# Plotting
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Plot loss function over epochs
ax[0].plot(range(epochs), loss_history, label="Loss", color="blue")
ax[0].set_title("Loss Function")
ax[0].set_xlabel("Epoch")
ax[0].set_ylabel("Loss")
ax[0].legend()

# Plot theta trajectory
theta_range = np.linspace(0, 5, 100)
loss_curve = (theta_range - 3) ** 2  # Loss curve
ax[1].plot(theta_range, loss_curve, label="Loss Curve", color="gray", linestyle="--")
ax[1].scatter(theta_history, loss_history, color="red", label="SGD Steps", zorder=5)
ax[1].set_title("SGD Trajectory")
ax[1].set_xlabel("Theta")
ax[1].set_ylabel("Loss")
ax[1].legend()

plt.tight_layout()
plt.show()
```
